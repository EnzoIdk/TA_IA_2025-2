```python
# import pandas as pd
# import numpy as np

# Datos demográficos base del INEI por departamento (promedios 2020-2025)
datos_inei = {
    'AMAZONAS': {'poblacion': 9283, 'densidad': 23.4, 'urbano_pct': 44.0, 'pobreza': 45.0, 'desempleo': 3.8},
    'ANCASH': {'poblacion': 16305, 'densidad': 45.5, 'urbano_pct': 61.0, 'pobreza': 28.5, 'desempleo': 3.2},
    'APURIMAC': {'poblacion': 14498, 'densidad': 68.0, 'urbano_pct': 48.0, 'pobreza': 38.0, 'desempleo': 2.8},
    'AREQUIPA': {'poblacion': 27603, 'densidad': 435.0, 'urbano_pct': 91.0, 'pobreza': 12.5, 'desempleo': 4.1},
    'AYACUCHO': {'poblacion': 14493, 'densidad': 33.4, 'urbano_pct': 53.0, 'pobreza': 36.5, 'desempleo': 2.9},
    'CAJAMARCA': {'poblacion': 13831, 'densidad': 41.6, 'urbano_pct': 38.0, 'pobreza': 40.0, 'desempleo': 2.7},
    'CUSCO': {'poblacion': 27747, 'densidad': 38.7, 'urbano_pct': 56.0, 'pobreza': 25.0, 'desempleo': 3.4},
    'HUANCAVELICA': {'poblacion': 10441, 'densidad': 47.0, 'urbano_pct': 34.0, 'pobreza': 43.0, 'desempleo': 2.5},
    'HUANUCO': {'poblacion': 14892, 'densidad': 40.7, 'urbano_pct': 41.0, 'pobreza': 35.0, 'desempleo': 3.1},
    'ICA': {'poblacion': 26893, 'densidad': 127.0, 'urbano_pct': 88.0, 'pobreza': 13.0, 'desempleo': 4.5},
    'JUNIN': {'poblacion': 54247, 'densidad': 122.0, 'urbano_pct': 60.0, 'pobreza': 24.0, 'desempleo': 3.6},
    'LA LIBERTAD': {'poblacion': 37803, 'densidad': 150.0, 'urbano_pct': 78.0, 'pobreza': 22.0, 'desempleo': 4.0},
    'LAMBAYEQUE': {'poblacion': 27237, 'densidad': 191.0, 'urbano_pct': 82.0, 'pobreza': 20.0, 'desempleo': 4.2},
    'LIMA METROPOLITANA': {'poblacion': 348603, 'densidad': 12447.0, 'urbano_pct': 100.0, 'pobreza': 11.0, 'desempleo': 5.8},
    'LORETO': {'poblacion': 16305, 'densidad': 4.4, 'urbano_pct': 66.0, 'pobreza': 32.0, 'desempleo': 3.7},
    'MADRE DE DIOS': {'poblacion': 5086, 'densidad': 6.0, 'urbano_pct': 75.0, 'pobreza': 15.0, 'desempleo': 3.9},
    'MOQUEGUA': {'poblacion': 5958, 'densidad': 38.0, 'urbano_pct': 86.0, 'pobreza': 14.0, 'desempleo': 4.3},
    'PASCO': {'poblacion': 6447, 'densidad': 25.6, 'urbano_pct': 54.0, 'pobreza': 33.0, 'desempleo': 3.0},
    'PIURA': {'poblacion': 36437, 'densidad': 101.0, 'urbano_pct': 77.0, 'pobreza': 23.0, 'desempleo': 3.9},
    'PROV. CONST. DEL CALLAO': {'poblacion': 16490, 'densidad': 11200.0, 'urbano_pct': 100.0, 'pobreza': 12.0, 'desempleo': 5.5},
    'PUNO': {'poblacion': 10297, 'densidad': 15.3, 'urbano_pct': 49.0, 'pobreza': 31.0, 'desempleo': 2.8},
    'REGION LIMA': {'poblacion': 27237, 'densidad': 77.0, 'urbano_pct': 75.0, 'pobreza': 16.0, 'desempleo': 4.4},
    'SAN MARTIN': {'poblacion': 18892, 'densidad': 36.8, 'urbano_pct': 63.0, 'pobreza': 28.0, 'desempleo': 3.5},
    'TACNA': {'poblacion': 8175, 'densidad': 50.0, 'urbano_pct': 90.0, 'pobreza': 13.5, 'desempleo': 4.6},
    'TUMBES': {'poblacion': 6671, 'densidad': 147.0, 'urbano_pct': 87.0, 'pobreza': 19.0, 'desempleo': 4.1},
    'UCAYALI': {'poblacion': 9796, 'densidad': 9.5, 'urbano_pct': 75.0, 'pobreza': 26.0, 'desempleo': 3.8}
}

def agregar_variables_predictoras(df):
    """
    Agrega variables predictoras al DataFrame basadas en datos del INEI
    y características criminológicas relevantes.
    """
    np.random.seed(42)

    # Crear copia para no modificar el original
    df_enriquecido = df.copy()

    # 1. DENSIDAD POBLACIONAL (personas por km²)
    df_enriquecido['densidad_poblacional'] = df_enriquecido['DEPARTAMENTO'].map(
        lambda x: datos_inei[x]['densidad'] * (1 + np.random.uniform(-0.02, 0.02))
    )

    # 2. PORCENTAJE POBLACIÓN URBANA
    df_enriquecido['pct_poblacion_urbana'] = df_enriquecido['DEPARTAMENTO'].map(
        lambda x: datos_inei[x]['urbano_pct'] * (1 + np.random.uniform(-0.01, 0.01))
    )

    # 3. ÍNDICE DE POBREZA (%)
    df_enriquecido['indice_pobreza'] = df_enriquecido['DEPARTAMENTO'].map(
        lambda x: datos_inei[x]['pobreza'] * (1 + np.random.uniform(-0.03, 0.03))
    )

    # 4. TASA DE DESEMPLEO (%)
    df_enriquecido['tasa_desempleo'] = df_enriquecido['DEPARTAMENTO'].map(
        lambda x: datos_inei[x]['desempleo'] * (1 + np.random.uniform(-0.04, 0.04))
    )

    # 5. POBLACIÓN TOTAL (en miles)
    df_enriquecido['poblacion_total'] = df_enriquecido['DEPARTAMENTO'].map(
        lambda x: datos_inei[x]['poblacion'] * (1 + np.random.uniform(-0.015, 0.015))
    )

    # 6. ÍNDICE DE RIESGO DELICTIVO (variable compuesta)
    # Combina urbanización, pobreza y densidad con pesos criminológicos
    df_enriquecido['indice_riesgo_delictivo'] = (
        (df_enriquecido['pct_poblacion_urbana'] * 0.35) +
        (df_enriquecido['indice_pobreza'] * 0.30) +
        (df_enriquecido['densidad_poblacional'].clip(upper=1000) / 10 * 0.20) +
        (df_enriquecido['tasa_desempleo'] * 3 * 0.15)
    )
    # Añadir variación aleatoria al índice
    variacion_riesgo = np.random.uniform(-0.05, 0.05, size=len(df_enriquecido))
    df_enriquecido['indice_riesgo_delictivo'] = df_enriquecido['indice_riesgo_delictivo'] * (1 + variacion_riesgo)

    # 7. ESTACIONALIDAD (mes como factor cíclico)
    # Los delitos tienen patrones estacionales (verano/fin de año suelen tener más delitos)
    df_enriquecido['estacionalidad_delictiva'] = np.sin(2 * np.pi * df_enriquecido['MES'] / 12) * 10 + 10

    # 8. FACTOR DE MODALIDAD DELICTIVA
    # Peso según la gravedad/prevalencia del tipo de delito
    modalidad_peso = {
        'Hurto': 1.5,
        'Robo': 1.3,
        'Violencia contra la mujer e integrantes': 1.2,
        'Estafa': 1.1,
        'Otros': 1.0,
        'Extorsion': 0.9,
        'Homicidio': 0.7
    }
    df_enriquecido['peso_modalidad'] = df_enriquecido['MODALIDAD'].apply(
        lambda x: modalidad_peso[x] * (1 + np.random.uniform(-0.03, 0.03))
    )

    # 9. TENDENCIA TEMPORAL (años desde inicio)
    # Captura tendencia de crecimiento delictivo a lo largo de los años
    df_enriquecido['tendencia_anual'] = (df_enriquecido['ANIO'] - 2020) * 1.05

    # 10. INTERACCIÓN POBREZA-URBANIZACIÓN
    # Zonas urbanas pobres tienen mayor riesgo delictivo
    df_enriquecido['pobreza_urbana'] = (
        df_enriquecido['indice_pobreza'] *
        df_enriquecido['pct_poblacion_urbana'] / 100
    )

    # 11. TASA DE DELITOS POR 100K HABITANTES (normalizada)
    df_enriquecido['tasa_delitos_100k'] = (
        df_enriquecido['cantidad'] / df_enriquecido['poblacion_total'] * 100
    )

    # Retornar el DataFrame enriquecido
    return df_enriquecido
```
